```python

import requests
import pandas as pd

# Official NCCC Open API
api_url = (
    "https://bas.nccc.com.tw/nccc-nop/"
    "OpenAPI/A08/ATMCashAdvanceTransaction"
)

# Request data
try:
    response = requests.get(api_url, timeout=30)
    response.raise_for_status()

    print("HTTP Status:", response.status_code)
    print("Content Type:", response.headers.get("Content-Type"))
    print("\nResponse preview:")
    print(response.text[:1000])

except requests.RequestException as e:
    print("API request failed:", e)
```

```python

from io import StringIO

# Parse the CSV data returned by the API
api_df = pd.read_csv(StringIO(response.content.decode("utf-8-sig")))

# Rename columns to English
api_df.columns = [
    "year",
    "cash_advance_amount_100m_twd",
    "cash_advance_transactions_1000"
]

# Convert columns to numeric values
for col in api_df.columns:
    api_df[col] = pd.to_numeric(api_df[col], errors="raise")

# Add data source information
api_df["source"] = "NCCC Taiwan"

# Display results
print("Dataset shape:", api_df.shape)
display(api_df.head())
```

```python
api_df.to_csv(
    "credit_card_api.csv",
    index=False,
    encoding="utf-8-sig"
)

print("CSV exported successfully!")
```
